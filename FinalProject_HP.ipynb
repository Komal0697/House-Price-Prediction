```python
import numpy  as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler,OneHotEncoder
```

```python
# Load dataset

df=pd.read_csv("HP_Cleaned_data.csv")
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
# seperate features and target

x=df.drop("Price",axis =1)
y=df["Price"]
```

```python
# encode categorical colunms
from sklearn.compose import ColumnTransformer

cat_cols =["Location","Property_Type"]
preprocessor =ColumnTransformer(transformers=[("cat",OneHotEncoder(handle_unknown="ignore"),cat_cols)],remainder ="passthrough")
```

```python
# Train Test Split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

# Random Forest Regression

```python
# Train Model

rf =RandomForestRegressor(n_estimators =100)

x_train_rf =preprocessor.fit_transform(x_train)
x_test_rf =preprocessor.transform(x_test)


rf.fit(x_train_rf,y_train)
```

```python
# Make Predictions

y_pred_rf = rf.predict(x_test_rf)
```

```python
# Evaluation of Model

from sklearn.linear_model import LinearRegression ,Ridge,Lasso


mae =mean_absolute_error(y_test,y_pred_rf)
mse =mean_squared_error(y_test,y_pred_rf)
rmse= np.sqrt(mse)
r2_score= r2_score(y_test,y_pred_rf)

print("Random Forest")
print("MAE:",mae)
print("MSE:",mse)
print("RMSE:",rmse)
print("r2 Score:",r2_score)
```

```python
# Insight:
Random forest combines multiple decision trees,helping to identify complex relationships and reduce the overfitting.
```

# AdaBoost Regression

```python
from sklearn.ensemble import AdaBoostRegressor

ada_model =AdaBoostRegressor(n_estimators=100)

ada_model.fit(x_train_rf,y_train)
```

```python
# Make Prediction

y_pred_ada =ada_model.predict(x_test_rf)
```

```python
# Evaluation Model

from sklearn.metrics import r2_score

ada_mae =mean_absolute_error(y_test,y_pred_ada)
ada_mse =mean_squared_error(y_test,y_pred_ada)
ada_rmse= np.sqrt(mse)
ada_r2 = r2_score(y_test,y_pred_ada)

print("AdaBoost Regression Model")
print("MAE:",ada_mae)
print("MSE:",ada_mse)
print("RMSE:",ada_rmse)
print("R2 Score:",ada_r2)
```

```python
# Insight:
Adaboost builds models sequentially and focuses more on observations that were predicted poorly by previous model.
```

# Gradient Boosting Regression

```python
from sklearn.ensemble import GradientBoostingRegressor

gb_model= GradientBoostingRegressor(n_estimators=100,max_depth=3)

gb_model.fit(x_train_rf,y_train)

```

```python
# Make Predictions

y_pred_gb =gb_model.predict(x_test_rf)
```

```python
# Evaluation Model 

from sklearn.metrics import r2_score
gb_mae =mean_absolute_error(y_test,y_pred_gb)
gb_mse =mean_squared_error(y_test,y_pred_gb)
gb_rmse= np.sqrt(gb_mse)
gb_r2= r2_score(y_test,y_pred_gb)

print("Gradient Boosting Rgression Model")
print("MAE:",gb_mae)
print("MSE:",gb_mse)
print("RMSE:",gb_rmse)
print("r2 Score:",gb_r2)
```

```python
# Insight:
Gradient Boosting builds trees sequentially,where each new tree attempts to correct the errors of the previous trees.
```
